```python
import os
import tensorflow as tf
assert tf.__version__.startswith('2')

from mediapipe_model_maker import gesture_recognizer

import matplotlib.pyplot as plt
%matplotlib inline
```

```python
dataset_path = "Personal_data_2"
```

```python
print(dataset_path)
labels = []
for i in os.listdir(dataset_path):
  if os.path.isdir(os.path.join(dataset_path, i)):
    labels.append(i)
print(labels)
```

```python
NUM_EXAMPLES = 5

for label in labels:
  label_dir = os.path.join(dataset_path, label)
  example_filenames = os.listdir(label_dir)[:NUM_EXAMPLES]
  fig, axs = plt.subplots(1, NUM_EXAMPLES, figsize=(10,2))
  for i in range(NUM_EXAMPLES):
    axs[i].imshow(plt.imread(os.path.join(label_dir, example_filenames[i])))
    axs[i].get_xaxis().set_visible(False)
    axs[i].get_yaxis().set_visible(False)
  fig.suptitle(f'Showing {NUM_EXAMPLES} examples for {label}')

plt.show()
```

```python
data = gesture_recognizer.Dataset.from_folder(
    dirname=dataset_path,
    hparams=gesture_recognizer.HandDataPreprocessingParams()
)
train_data, rest_data = data.split(0.8)
validation_data, test_data = rest_data.split(0.5)
```

```python
hparams = gesture_recognizer.HParams(learning_rate=0.001,
                                     export_dir="Model_w_Brightness_200",
                                     epochs=200,
                                     shuffle=True,
                                     batch_size=64)
# model_options = gesture_recognizer.ModelOptions(layer_widths=[64, 128, 256])
options = gesture_recognizer.GestureRecognizerOptions(hparams=hparams)
model = gesture_recognizer.GestureRecognizer.create(
    train_data=train_data,
    validation_data=validation_data,
    options=options
)
```

```python
loss, acc = model.evaluate(test_data, batch_size=16)
print(f"Test loss:{loss}, Test accuracy:{acc}")
```

```python
import cv2

import math

DESIRED_HEIGHT = 480
DESIRED_WIDTH = 480

def resize_and_show(image):
  h, w = image.shape[:2]
  if h < w:
    img = cv2.resize(image, (DESIRED_WIDTH, math.floor(h/(w/DESIRED_WIDTH))))
  else:
    img = cv2.resize(image, (math.floor(w/(h/DESIRED_HEIGHT)), DESIRED_HEIGHT))
  cv2.imshow("Resized Image", img)  # Corrected line
  # cv2.waitKey(1000)  # Add this line if you want to wait until a key is pressed
  cv2.destroyAllWindows() 


# Preview the images.
images = {name: cv2.imread(name) for name in IMAGE_FILENAMES}
for name, image in images.items():
  print(name)
  resize_and_show(image)
```

```python
# STEP 1: Import the necessary modules.
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# STEP 2: Create an GestureRecognizer object.
base_options = python.BaseOptions(model_asset_path='/mnt/Main Drive/Codes/Deep Learning/Gesture_control/Mediapipe_Gesture/Model_w_Brightness_200/gesture_recognizer.task')
options = vision.GestureRecognizerOptions(base_options=base_options)
recognizer = vision.GestureRecognizer.create_from_options(options)

images = []
results = []
for image_file_name in IMAGE_FILENAMES:
  # STEP 3: Load the input image.
  image = mp.Image.create_from_file(image_file_name)

  # STEP 4: Recognize gestures in the input image.
  recognition_result = recognizer.recognize(image)

  # STEP 5: Process the result. In this case, visualize it.
  images.append(image)
  if recognition_result.gestures:  # Check if the list is not empty
    top_gesture = recognition_result.gestures[0][0].category_name
    print(top_gesture)
    hand_landmarks = recognition_result.hand_landmarks
    results.append((top_gesture, hand_landmarks))
  else:
    print("No gestures recognized in this image.")
    results.append(("No gestures recognized", None))

# print(results[2][0])
# cv2.imshow(results)
```
